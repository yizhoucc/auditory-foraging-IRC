```python
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
from auditoryforage.utils import open_pickle_file
folder_list_keys = ['100-500', '500-1000','1000-1500','1500-2000']
file_names_list = ['../' + folder_list_key + '/store/ft_data.pkl' for folder_list_key in folder_list_keys]
file_name = './store/ft_data.pkl'

master_ft_data = {}
for file_name in file_names_list:
    ft_data = open_pickle_file(file_name)
    master_ft_data.update(ft_data)
```

```python
plt.figure(figsize=(10, 6))
options = ['no_normalized_offset', 'no_normalized_no_offset', 'normalized_offset', 'normalized_no_offset']
choice = 1
for fr_val in master_ft_data.keys():
    plt.plot(master_ft_data[fr_val][options[choice]]['freq'], master_ft_data[fr_val][options[choice]]['mag'], label = 'food reward ' + str(fr_val))
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.title('FT ' +  options[choice])
plt.legend()
plt.grid(True)
plt.show()
```
