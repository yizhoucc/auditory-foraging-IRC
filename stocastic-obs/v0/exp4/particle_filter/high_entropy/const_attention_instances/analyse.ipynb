```python
import matplotlib.pyplot as plt
import numpy as np
import random
from itertools import permutations

from auditoryforage.utils import open_pickle_file, save_pickle_file
file_name = './c_0.6-0.9_t_0.4-0.9.pkl'
# file_name = './simulation_results.pkl'
dict = open_pickle_file(file_name)
```

```python
certainity_list = list(dict.keys())[1:]
thresh_list = list(dict[0.6].keys())
for certainity in certainity_list:
    plt.plot(thresh_list, [np.max(dict[certainity][thresh]) for thresh in thresh_list],'*-', label = f'{certainity}')
plt.xlabel('threshold')
plt.ylabel('max avg. reward')
legend = plt.legend()
legend.set_title('Certainity')
plt.show()  
```
