```python
%matplotlib inline
import matplotlib.pyplot as plt
import os
import numpy as np
from auditoryforage.utils import save_pickle_file, open_pickle_file

folder_name = './store/filters/'
file_name = folder_name + 'PF_likelihood_repo.pkl'

if os.path.exists(file_name):
    PF_likelihood_repo = open_pickle_file(file_name)
else:
    print('Have not run previous cell yet.')

total_posterior = [0] * len(PF_likelihood_repo.keys())
for key in PF_likelihood_repo:
    _, correct_env_index = key
    PF_posterior_full_array = np.array(PF_likelihood_repo[key])
    plt.figure()
    plt.errorbar(np.arange(len(PF_likelihood_repo.keys())), np.mean(PF_posterior_full_array, axis=0), yerr=np.std(PF_posterior_full_array, axis=0), fmt='o', markersize=6, capsize=4)
    plt.xlabel('environment index')
    plt.ylabel('normalized count')
    plt.title(f'Correct index is {correct_env_index}')
    plt.xticks(np.arange(len(PF_likelihood_repo.keys())))  # Set x-axis labels to column indices
    plt.grid()
    plt.show()
```
