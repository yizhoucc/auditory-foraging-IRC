```python
import matplotlib.pyplot as plt
from auditoryforage.utils import open_pickle_file

main_folder_name = './prob_0.85_ent_'
folder_sub_names = ['0', '0.001', '.01', '0.1']
file_sub_dir = '/store/comparison/diff_food_rewards.pkl'
legends_list = []
line_styles = ['-.','--', ':',  '-']
for folder_sub_name in folder_sub_names:
    detection_measures = open_pickle_file(main_folder_name + folder_sub_name + file_sub_dir)
    food_reward_list = detection_measures['input']['food_reward_list']
    hit_count_list = detection_measures['output']['hit_count_list']
    false_alarm_list = detection_measures['output']['false_alarm_list']
    miss_count_list = detection_measures['output']['miss_count_list']
    line_style = line_styles.pop()
    plt.plot(food_reward_list, hit_count_list, f'g{line_style}')
    plt.plot(food_reward_list, false_alarm_list, f'r{line_style}')
    plt.plot(food_reward_list, miss_count_list, f'b{line_style}')
    legends_list += ['hit ' + folder_sub_name, 'miss ' + folder_sub_name, 'false alarm ' + folder_sub_name]
plt.xlabel('food_reward_list')
plt.legend(legends_list)
plt.show()
```
