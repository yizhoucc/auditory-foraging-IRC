## Run and save particle filters' outputs in parallel

```python
from auditoryforage.utils import particle_filter
no_particles = 500
state_list = [list_of_state[0] for list_of_state in episode['states']]
lick_actions = [env.dict_action_possible[action][0] for action in episode['actions']]

reference_episode = episode

number_of_runs = 5
file_names = [f'PF_IO_episodes_run_{ind}.pkl' for ind in range(number_of_runs)]

def run_and_save_particle_filter(file_name):
    particle_filter_output = particle_filter(agent = agent, env = env, lick_actions = lick_actions, state_list = state_list, no_particles = no_particles, sampling_freq = 1)
    particle_filter_compare_data = {'reference_episode': episode, "particle_filter_output": particle_filter_output}
    open_file = open(file_name, "wb")
    pickle.dump(particle_filter_compare_data, open_file)
    open_file.close()
    print(f'Saved file {file_name}', flush=True)

with multiprocessing.Pool() as pool:
    pool.map(run_and_save_particle_filter, file_names)
```

### Checking if there are at least two particles with different trajectories.

```python
def check_for_different_trajectories(particle_filter_compare_data, verbose = False):
    matching_indices = []
    need_no_particles = False
    did_it_converge = False
    try: no_particles
    except NameError: need_no_particles = True
    if not need_no_particles:
        ind = 0
        diff_in_actions = 0
        while diff_in_actions == 0:
            ind += 1
            if ind > no_particles - 1:
                did_it_converge = True
                if verbose: print('All particles converged to the same trajectory.')
                break
            a = particle_filter_compare_data['output']['generated_episodes'][0]['actions']
            b = particle_filter_compare_data['output']['generated_episodes'][ind]['actions']
            diff_in_actions = np.sum(np.abs(a-b))/len(a)
        if not did_it_converge:
            matching_indices.append(0)
            matching_indices.append(ind)
            if verbose:
                print(f'Found at least two particles with different trajectories, namely particle {ind} and particle 0.')
                print(f'Rate of difference in actions between them is {diff_in_actions} per time step.')
    else:
        print('Provide number of particles before running this code!')
    return matching_indices
_ = check_for_different_trajectories(particle_filter_compare_data, verbose = True)
```

### Checking if there are at least two particles with different trajectories withing a PF run, for all runs of PF.

```python
list_of_matching_indices = []
PF_runs_outputs = {}
verbose = True

for file_name in file_names:
    open_file = open(file_name, "rb")
    particle_filter_compare_data = pickle.load(open_file)
    open_file.close()
    if verbose: print(f'\n Checking file {file_name}')
    list_of_matching_indices.append(check_for_different_trajectories(particle_filter_compare_data, verbose = verbose))
    PF_runs_outputs[file_name] = particle_filter_compare_data
    if len(list_of_matching_indices) == 0:
        PF_runs_outputs[file_name]["particle_filter_output"] = particle_filter_compare_data

    PF_runs_outputs[file_name] = particle_filter_compare_data
```

### Visually comparing reference and generated episode

```python
def plot_comparison(reference_ep, generated_ep, start, stop):
    def tranform_beleifs(episode):
        return np.log(episode['q_probs'].T[:,start:stop])

    def tranform_actions(episode):
        return episode['actions'].T[start:stop]

    plt.subplot(2,1,1)
    plt.imshow(tranform_beleifs(reference_ep))
    plt.colorbar()
    plt.xlabel('time')
    plt.ylabel('state')
    plt.subplot(2,1,2)
    plt.imshow(tranform_beleifs(generated_ep))
    plt.xlabel('time')
    plt.ylabel('state')
    plt.colorbar()
    plt.show()

    plt.figure()
    plt.subplot(2,1,1)
    plt.stem(tranform_actions(reference_ep))
    plt.xlabel('time')
    plt.ylabel('action')
    plt.subplot(2,1,2)
    plt.stem(tranform_actions(generated_ep))
    plt.xlabel('time')
    plt.ylabel('action')
    plt.show()
```

```python
file_name = 'store/particle_filter/PF_np_100_sf_10_19.pkl'
open_file = open(file_name, "rb")
particle_filter_IO = pickle.load(open_file)
open_file.close()
reference_ep = particle_filter_IO['input']['root_episode']
generated_ep = particle_filter_IO['output']['generated_episodes'][0]
plot_comparison(reference_ep, generated_ep, start= 0, stop= len(generated_ep['actions']))
```
