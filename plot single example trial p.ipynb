```python
# obs result
res=[]
shading=0.3
for o,a in zip(obs, att):
    # o=o[0]
    # print(o,a)
    if a==1: 
        if o==0: res.append(-1)
        else: res.append(1)
    elif a==0:
        if o==0: res.append(-shading)
        else: res.append(shading)
res=np.array(res)
```
